```julia
using DataFrames, CSV
```

```julia
using Random, Flux
```

```julia
using Statistics
```

```julia
df = CSV.read("creditcard.csv", DataFrame)
```

```julia
X = Matrix{Float32}(df[:, Not(:Class)])'
```

```julia
y = reshape(Float32.(df.Class), 1,:)
```

```julia
sum(y)
```

```julia
size(X)
```

```julia
n = size(X, 2)
```

```julia
ntrain = floor(Int, 0.8*n)
```

```julia
indices = randperm(n)
```

```julia
train_idx = indices[1:ntrain]
```

```julia
test_idx = indices[ntrain+1:end]
```

```julia
Xtrain = X[:, train_idx]
Xtest = X[:, test_idx]
```

```julia
y_train = y[:, train_idx]
y_test = y[:, test_idx]
```

```julia
μ = mean(Xtrain, dims=2)
```

```julia
σ = std(Xtrain, dims=2)
```

```julia
Xtrain = (Xtrain .- μ) ./ σ
Xtest = (Xtest .- μ) ./  σ
```

```julia
model = Chain(Dense(30, 128, relu), Dense(128, 32, relu), Dense(32, 1), sigmoid)
```

```julia
n0 = sum(y_train .== 0)
```

```julia
n1 = sum(y_train .== 1)
```

```julia
N = length(y_train)
```

```julia
w0 = Float32(N / (2*n0))
```

```julia
w1 = Float32(N / (2*n1))
```

```julia
function loss(model, x, y) 
    prob = model(x)
    weights = ifelse.(y .== 1f0, w1, w0)
    return mean(weights .* (.-y .* log.(prob .+ 1f-7) .- (1f0 .- y).*log.(1f0 .- prob .+ 1f-7)))
end
```

```julia
optimizer = Flux.setup(Flux.Adam(0.001), model)
```

```julia
for epoch=1:100
    Flux.train!(loss, model, [(Xtrain, y_train)], optimizer)
    println("Epoch $epoch loss :", loss(model, Xtrain, y_train))
end
```

```julia
probabilities = model(Xtest)
```

```julia
predictions = Float32.(probabilities .>= 0.5)
```

```julia
tp = sum((predictions .== 1) .& (y_test .== 1))
```
